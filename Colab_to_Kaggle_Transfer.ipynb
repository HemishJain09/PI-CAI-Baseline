# Push PI-CAI Dataset from Google Drive to Kaggle

This notebook will take your 26GB pre-processed PI-CAI dataset sitting in Google Drive and upload it directly to Kaggle as a **Private Dataset**. 

By running this on Colab, it uses Google's ultra-fast cloud internet to transfer the 26GB directly to Kaggle, so you don't have to download anything to your personal computer!

### Prerequisites:
1. Go to [Kaggle.com](https://www.kaggle.com/) -> Click your profile picture -> Settings
2. Scroll down to **API** and click **Create New Token**.
3. This will download a `kaggle.json` file to your computer. Keep it handy.

```python
# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. Upload Kaggle API Key
from google.colab import files
import os

print("Please upload the kaggle.json file you downloaded from Kaggle:")
uploaded = files.upload()

!mkdir -p ~/.kaggle
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json
print("Kaggle API key configured successfully!")
```

```python
# 3. Set up your Dataset Information
import json

# IMPORTANT: Change 'your_kaggle_username' to your actual Kaggle username!
KAGGLE_USERNAME = "your_kaggle_username"
DATASET_TITLE = "PI-CAI Pre-processed"
DATASET_SLUG = "picai-pre-processed-dataset" # Must be lowercase, alphanumeric, and hyphens only

metadata = {
  "title": DATASET_TITLE,
  "id": f"{KAGGLE_USERNAME}/{DATASET_SLUG}",
  "licenses": [
    {
      "name": "CC0-1.0"
    }
  ]
}

# Define the path to your data in Google Drive
DATA_FOLDER = "/content/drive/MyDrive/PI-CAI_pre-processed"

# Write the metadata file into the folder so Kaggle knows what to upload
with open(f"{DATA_FOLDER}/dataset-metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

print("Metadata created successfully!")
```

```python
# 4. Push the Dataset to Kaggle
# This step will take some time because it is transferring ~26GB of data.
# Sit back and relax until it finishes!

!kaggle datasets create -p "$DATA_FOLDER" --dir-mode zip
```

### What to do next?

Once the upload finishes successfully:
1. Go to Kaggle and create a New Notebook.
2. On the right-hand sidebar, click **Add Data** -> **Your Datasets** and add `PI-CAI Pre-processed`.
3. You can now access your data instantly at `/kaggle/input/picai-pre-processed-dataset` at ultra-high speeds!
